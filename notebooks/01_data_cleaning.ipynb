- **product_category_name**

- **product_category_name_english**  
  Ürün kategorisinin İngilizce karşılığını gösterir.

Bu veri seti ürün kategorilerini daha anlaşılır hale getirmek için kullanılır.  
Products veri seti ile birleştirilerek kategorilerin İngilizce isimleri üzerinden analiz yapılabilir.

```python
# Temel tip dönüşümleri ve temizleme

# Tarih kolonlarını datetime tipine çevir
date_cols_orders = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

for col in date_cols_orders:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

order_reviews["review_creation_date"] = pd.to_datetime(order_reviews["review_creation_date"], errors="coerce")
order_reviews["review_answer_timestamp"] = pd.to_datetime(order_reviews["review_answer_timestamp"], errors="coerce")

# Boş product kategorilerini "unknown" ile doldur
products["product_category_name"] = products["product_category_name"].fillna("unknown")

# Ürün ölçüleri ve ağırlığındaki eksik verileri medyan ile doldur
for col in ["product_weight_g", "product_length_cm", "product_height_cm", "product_width_cm"]:
    products[col] = products[col].fillna(products[col].median())

# Negatif fiyat veya kargo ücreti varsa filtrele
order_items = order_items[(order_items["price"] >= 0) & (order_items["freight_value"] >= 0)]

# Aykırı değer (Outlier) temizliği (Basit IQR Yöntemi ile aşırı yüksek fiyatları filtreleme)
Q1 = order_items["price"].quantile(0.25)
Q3 = order_items["price"].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 3 * IQR # 3 IQR kullanarak daha toleranslı bir filtreleme yapıyoruz
order_items = order_items[order_items["price"] <= upper_bound]

# --- Geolokasyon Verisinin Hazırlanması ve Birleştirilmesi ---
# Aynı posta kodu için birden fazla enlem/boylam olabildiği için ortalamasını alıyoruz
geo_grouped = geolocation.groupby("geolocation_zip_code_prefix", as_index=False).agg(
    geolocation_lat=("geolocation_lat", "mean"),
    geolocation_lng=("geolocation_lng", "mean")
)

customers = customers.merge(
    geo_grouped, 
    left_on="customer_zip_code_prefix", 
    right_on="geolocation_zip_code_prefix", 
    how="left"
).drop(columns=["geolocation_zip_code_prefix"])

sellers = sellers.merge(
    geo_grouped, 
    left_on="seller_zip_code_prefix", 
    right_on="geolocation_zip_code_prefix", 
    how="left",
    suffixes=("", "_seller")
).drop(columns=["geolocation_zip_code_prefix"])

# --- Ana fact tabloyu oluşturma ---

# orders + customers birleştirme
orders_cust = orders.merge(
    customers,
    on="customer_id",
    how="left",
    suffixes=("", "_customer"),
)

# order_payments üzerinden sipariş bazında toplam ödeme bilgisi
payments_agg = (
    order_payments
    .groupby("order_id", as_index=False)
    .agg(
        payment_value=("payment_value", "sum"),
        payment_installments=("payment_installments", "max"),
    )
)

orders_cust_pay = orders_cust.merge(payments_agg, on="order_id", how="left")

# order_items + products birleştirme
items_prod = order_items.merge(products, on="product_id", how="left")

# Ürün kategorisinin İngilizce karşılığını ekle
items_prod = items_prod.merge(
    category_translation,
    on="product_category_name",
    how="left",
)

# Satıcı bilgilerini ekle
items_prod_seller = items_prod.merge(
    sellers,
    on="seller_id",
    how="left",
    suffixes=("", "_seller"),
)

# Sipariş bazlı tablo ile satır bazlı ürün tablosunu birleştir
fact_orders = items_prod_seller.merge(
    orders_cust_pay,
    on="order_id",
    how="left",
    suffixes=("", "_order"),
)

# Yorum verilerini sipariş bazında özetleyip ekle
reviews_agg = (
    order_reviews
    .groupby("order_id", as_index=False)
    .agg(
        review_score=("review_score", "mean"),
        review_count=("review_id", "count"),
    )
)

fact_orders = fact_orders.merge(reviews_agg, on="order_id", how="left")

# Yardımcı kolonlar (metrikler için tarih alanları ve gelir)
fact_orders["total_item_value"] = fact_orders["price"] + fact_orders["freight_value"]

fact_orders["order_date"] = fact_orders["order_purchase_timestamp"].dt.date
fact_orders["order_year"] = fact_orders["order_purchase_timestamp"].dt.year
fact_orders["order_month"] = fact_orders["order_purchase_timestamp"].dt.to_period("M").astype(str)

# Temizlenmiş verileri kaydet
os.makedirs(DATA_PROCESSED, exist_ok=True)

fact_orders.to_parquet(os.path.join(DATA_PROCESSED, "fact_orders.parquet"), index=False)
orders_cust_pay.to_parquet(os.path.join(DATA_PROCESSED, "orders_customers_payments.parquet"), index=False)
products.to_parquet(os.path.join(DATA_PROCESSED, "products_clean.parquet"), index=False)
customers.to_parquet(os.path.join(DATA_PROCESSED, "customers_clean.parquet"), index=False)
```
